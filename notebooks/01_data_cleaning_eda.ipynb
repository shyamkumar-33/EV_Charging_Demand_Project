```python

# ============================================================
# FILE 1: DATA CLEANING AND EXPLORATORY DATA ANALYSIS (EDA)
# EV Charging Demand Prediction Project
# ============================================================

# STEP 1: INSTALL REQUIRED LIBRARIES
# Run this cell using the notebook's Python kernel.

%pip install pandas numpy matplotlib seaborn scikit-learn ipykernel

# STEP 2: IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

print("All required libraries imported successfully!")

# STEP 3: FIND THE PROJECT FOLDER

ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

DATA_PATH = ROOT / "data" / "charging_records.csv"

print("Project folder:", ROOT)
print("Dataset path:", DATA_PATH)

# STEP 4: LOAD THE DATASET

if not DATA_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found at: {DATA_PATH}\n"
        "Please put charging_records.csv inside the data folder."
    )

df = pd.read_csv(DATA_PATH)

print("\nDataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df.head())

# STEP 5: INSPECT THE DATASET

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe(include="all").T)

# STEP 6: CHECK MISSING VALUES

print("\nMissing values in each column:")
display(df.isnull().sum().sort_values(ascending=False))

# STEP 7: CHECK DUPLICATES

print("\nDuplicate rows:", df.duplicated().sum())

# Remove exact duplicate rows
df = df.drop_duplicates().copy()

print("Shape after removing exact duplicates:", df.shape)

# STEP 8: CLEAN COLUMN NAMES

df.columns = df.columns.str.strip()

# STEP 9: CONVERT DATE AND NUMERIC COLUMNS

for col in ["StartDatetime", "EndDatetime"]:
    if col in df.columns:
        df[col] = pd.to_datetime(
            df[col], errors="coerce"
        )

if "Demand" in df.columns:
    df["Demand"] = pd.to_numeric(
        df["Demand"], errors="coerce"
    )

if "Duration" in df.columns:
    df["Duration"] = pd.to_numeric(
        df["Duration"], errors="coerce"
    )

print("\nUpdated data types:")
print(df.dtypes)

# STEP 10: CHECK MISSING VALUES AFTER CONVERSION

print("\nMissing values after type conversion:")
display(df.isnull().sum().sort_values(ascending=False))

# STEP 11: REMOVE ROWS WITHOUT A VALID DEMAND VALUE

if "Demand" not in df.columns:
    raise KeyError(
        "The dataset has no 'Demand' column. "
        "Check the actual column names."
    )

df = df.dropna(subset=["Demand"]).copy()

print("\nRows remaining after removing missing Demand:", len(df))

# STEP 12: EXPLORATORY DATA ANALYSIS (EDA)

# 12A: Demand statistics
print("\nDemand statistics:")
display(df["Demand"].describe())

# 12B: Demand distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["Demand"], bins=30, kde=True)
plt.title("Distribution of EV Charging Demand")
plt.xlabel("Demand")
plt.ylabel("Number of Sessions")
plt.tight_layout()
plt.show()

# 12C: Demand by charger type
if "ChargerType" in df.columns:
    plt.figure(figsize=(9, 5))
    sns.boxplot(data=df, x="ChargerType", y="Demand")
    plt.title("Charging Demand by Charger Type")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()

# 12D: Average demand by location
if "Location" in df.columns:
    location_demand = (
        df.groupby("Location")["Demand"]
        .mean()
        .sort_values(ascending=False)
        .head(15)
    )

    location_demand.plot(
        kind="bar", figsize=(10, 5)
    )
    plt.title("Top 15 Locations by Average Demand")
    plt.xlabel("Location")
    plt.ylabel("Average Demand")
    plt.tight_layout()
    plt.show()

# STEP 13: FEATURE ENGINEERING

if "StartDatetime" in df.columns:
    df["Hour"] = df["StartDatetime"].dt.hour
    df["Day"] = df["StartDatetime"].dt.day
    df["Month"] = df["StartDatetime"].dt.month
    df["DayOfWeek"] = df["StartDatetime"].dt.dayofweek

    df["IsWeekend"] = (
        df["DayOfWeek"] >= 5
    ).astype(int)

    df["TimeOfDay"] = pd.cut(
        df["Hour"],
        bins=[-1, 5, 11, 17, 23],
        labels=[
            "Night",
            "Morning",
            "Afternoon",
            "Evening"
        ]
    )

    # Average demand by hour
    hourly_demand = df.groupby("Hour")["Demand"].mean()

    hourly_demand.plot(
        kind="line", marker="o", figsize=(9, 5)
    )
    plt.title("Average Charging Demand by Hour")
    plt.xlabel("Hour of Day")
    plt.ylabel("Average Demand")
    plt.xticks(range(24))
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# STEP 14: SAVE THE CLEANED DATASET

OUTPUT_PATH = ROOT / "data" / "cleaned_charging_data.csv"

df.to_csv(OUTPUT_PATH, index=False)

print("\n========================================")
print("DATA CLEANING AND EDA COMPLETED!")
print("Cleaned dataset saved to:")
print(OUTPUT_PATH)
print("Final dataset shape:", df.shape)
print("========================================")
```
